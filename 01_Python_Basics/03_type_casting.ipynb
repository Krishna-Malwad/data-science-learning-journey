# TYPE CASTING IN PYTHON

```python
# Integer to Float

a = 10

b = float(a)

print(b)
print(type(b))
```

```python
# Float to Integer

x = 10.9

y = int(x)

print(y)
print(type(y))
```

```python
# Integer to String

a = 100

b = str(a)

print(b)
print(type(b))
```

```python
# String to Integer

x = "50"

y = int(x)

print(y)
print(type(y))
```

```python
# Boolean Type Casting

print(bool(1))
print(bool(0))
print(bool("Python"))
print(bool(""))
```

```python
# VALID TYPE CONVERSIONS

# int to float
print(float(10))

# float to int
print(int(10.5))

# int to string
print(str(100))

# string num to int
print(int("50"))

# string to float
print(float("10.5"))

# int to bool
print(bool(1))

# empty string to bool
print(bool(""))
```
