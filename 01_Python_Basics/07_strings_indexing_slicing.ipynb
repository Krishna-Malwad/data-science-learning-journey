# STRINGS IN PYTHON

A string is a sequence of characters enclosed in quotes.

# RULES OF STRINGS

1. Strings are immutable (cannot be changed)
2. Can use single (' '), double (" "), triple quotes
3. Each character has an index starting from 0
4. Negative indexing starts from -1 (end of string)

```python
# Example of string

s = "Python"
print(s)
```

# INDEXING IN STRINGS
Indexing means accessing a single character from string.

# RULES OF INDEXING

1. Index starts from 0
2. Negative index starts from -1
3. Out of range gives error

```python
s = "Python"

print(s[0])   # first character
print(s[3])   # middle character
print(s[-1])  # last character
```

# SLICING IN STRINGS
Slicing means extracting part of string.

# RULES OF SLICING

string[start:end:step]

1. start is included
2. end is excluded
3. step is optional (default = 1)

```python
s = "Programming"

print(s[0:4])     # Prog
print(s[3:7])     # gram
print(s[:5])      # Progr
print(s[5:])      # amming
```

```python
# STEP IN SLICING

s = "Programming"

print(s[0:10:2])   # alternate characters
print(s[::-1])     # reverse string
```
