# BITWISE OPERATOR IN PYTHON

# WHAT ARE BITWISE OPERATORS?

Bitwise operators work on binary numbers (0 and 1).
They operate bit by bit.

# RULES OF BITWISE OPERATORS

1. Works only on integers
2. First converts number into binary
3. Then performs operation bit by bit
4. Final result is converted back to decimal
5. Used in low-level programming, AI, system design

# TYPES OF BITWISE OPERATORS

(&)  -> AND ||
(|)  -> OR ||
(^)  -> XOR ||
(~)  -> NOT ||
(<<) -> LEFT SHIFT ||
(>>) -> RIGHT SHIFT

# BITWISE AND (&)

Returns 1 if both bits are 1

```python
a = 5   # 0101
b = 3   # 0011

print(a & b)
```

# BITWISE OR (|)

Returns 1 if any one bit is 1

```python
a = 5
b = 3

print(a | b)
```

# BITWISE XOR (^)

Returns 1 if bits are different

```python
a = 5
b = 3

print(a ^ b)
```

# BITWISE NOT (~)

Flips all bits (0 → 1, 1 → 0)

```python
a = 5

print(~a)
```

# LEFT SHIFT (<<)

Shifts bits to left → multiplies number by 2

```python
a = 5

print(a << 1)
```

# RIGHT SHIFT (>>)

Shifts bits to right → divides number by 2

```python
a = 5

print(a >> 1)
```

# PRACTICE QUESTIONS
