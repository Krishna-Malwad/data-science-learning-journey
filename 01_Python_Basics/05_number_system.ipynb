# TYPES OF NUMBER SYSTEM

```python
# 1. Decimal (Base 10)
# 2. Binary (Base 2)
# 3. Octal (Base 8)
# 4. Hexadecimal (Base 16)
```

# DECIMAL NUMBER SYSTEM

```python
# Default number system in Python

a = 25
print(a)
```

# BINARY NUMBER SYSTEM

```python
# Prefix: 0b

b = 0b1010
print(b)
```

# OCTAL NUMBER SYSTEM

```python
# Prefix: 0o

o = 0o12
print(o)
```

# HEXADECIMAL NUMBER SYSTEM

```python
# Prefix: 0x

h = 0xA
print(h)
```

# PRACTICE QUESTIONS

```python
# 1. Convert binary to decimal
b = 0b1101
print(b)

# 2. Convert octal to decimal
o = 0o17
print(o)

# 3. Convert hexadecimal to decimal
h = 0x1A
print(h)
```

```python
# 4. Create your own binary, octal, hexadecimal numbers

bin_num = 0b111
oct_num = 0o10
hex_num = 0xF

print(bin_num)
print(oct_num)
print(hex_num)
```
