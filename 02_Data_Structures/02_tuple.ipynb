# TUPLE IN PYTHON

A tuple is a collection of items in Python that is similar to a list.

But the main difference is:
Tuple is IMMUTABLE (cannot be changed after creation).

Tuples are used when data should not be modified.

Example use cases:
- storing fixed data (like days of week)
- coordinates (x, y)
- database records that should not change

Tuples are faster than lists because they are immutable.

# RULES OF TUPLE

1. Ordered (index starts from 0)
2. Immutable (cannot be changed)
3. Allows duplicate values
4. Written using round brackets ()
5. Can store multiple data types
6. While declaring a single value in tuple always add a comma after that value otherwise system bydefault consider as int,float or any other data type depending on the value.

```python
# Creating Tuple
t1 = (10, 20, 30, "Python", True)
print(t1)
```

```python
# single element tuple (IMPORTANT RULE)
t2 = ('10')
print(type(t2))
```

```python
t3 = (1, 2, 3, 4)
print(t3[0])
print(t3[-1])
t3.
```

# TUPLE METHODS

In Tuple we have only two methods

# count()
Returns number of times a value appears in tuple / Returns occurences of value 

```python
t = (1, 2, 2, 3, 2)
print(t.count(2))
```

# index()
Returns index of first occurrence of value

```python
t = (10, 20, 30, 20)
print(t.index(20))
```

# LIST VS TUPLE

List:
- Mutable
- Uses square braces []

Tuple:
- Immutable
- Uses round braces ()
