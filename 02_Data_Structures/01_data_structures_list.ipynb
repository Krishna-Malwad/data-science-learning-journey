# DATA STRUCTURES IN PYTHON

Data structures are ways of storing and organizing data in a computer so that it can be used efficiently.

In real life, we deal with data in different formats:
- Student records
- Bank accounts
- Shopping lists
- Social media data

Python provides built-in data structures to manage this data easily.

Data structures help in:
- storing data
- accessing data
- modifying data
- managing large datasets efficiently

# TYPES OF DATA STRUCTURES

Data structures are mainly divided into two types:

1. Primitive Data Structures
   - int
   - float
   - string
   - bool

2. Non-Primitive Data Structures
   - List
   - Tuple
   - Set
   - Dictionary

# GENERAL RULES

1. Data structures are used to organize data
2. Each structure has its own behavior
3. Some are mutable (list, set)
4. Some are immutable (tuple)
5. Choice depends on use case

# LIST IN PYTHON

A list is a collection of items stored in a single variable.

It is one of the most flexible data structures in Python.

List can store:
- numbers
- strings
- boolean values
- even another list

Main feature of list:
It is MUTABLE (can be changed after creation)

Example use:
- storing marks of students
- storing names of users
- managing items in a shopping cart

# RULES OF LIST

1. Ordered (index starts from 0)
2. Mutable (can be changed)
3. Allows duplicate values
4. Can store multiple data types
5. Written using square brackets []

```python
#creating a list
# empty list
l1 = []
print(l1)
```

```python
# list with elements
l2 = [10, 20, 30, "Python", True]
print(l2)
```

```python
# accessing elements
l = [10, 20, 30, 40]

print(l[0])
print(l[-1])
```

# LIST FUNCTIONS (WITH EXAMPLES)

# append()
Adds element at end of list

```python
l = [1, 2, 3]
l.append(100)
print(l)
```

# insert()
Adds element at specific position

```python
l = [1, 2, 3]
l.insert(1, 999)
print(l)
```

# remove()
Removes first occurrence of value

```python
l = [10, 20, 30, 20]
l.remove(20)
print(l)
```

# pop()
Removes element by index (default last)

```python
l = [10, 20, 30]
l.pop()
print(l)
```

# sort()
Sorts list in ascending order

```python
l = [5, 2, 9, 1]
l.sort()
print(l)
```

# reverse()
Reverses list order

```python
l = [1, 2, 3, 4]
l.reverse()
print(l)
```

# clear()
Removes all elements from the list but keeps the list empty.

```python
l = [1, 2, 3]
l.clear()
print(l)
```

# copy()
Creates a shallow copy of the list. ||
Changes in copied list do not affect original list.

```python
l1 = [1, 2, 3]
l2 = l1.copy()

l2.append(100)

print(l1)
print(l2)
```

# count()
Returns how many times a value appears in the list.

```python
l = [1, 2, 2, 3, 2]

print(l.count(2))
```

# extend()
Adds multiple elements from another list into current list.

```python
l1 = [1, 2, 3]
l2 = [4, 5, 6]

l1.extend(l2)
print(l1)
```

# index()
Returns index of first occurrence of a value.

```python
l = [10, 20, 30, 20]

print(l.index(20))
```
