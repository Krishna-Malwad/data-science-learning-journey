# SET OPERATIONS IN PYTHON

Set operations are mathematical operations performed between two or more sets.

These operations are mainly used to:
- combine data
- compare data
- find common values
- remove duplicates

Python provides:
1. Operators
2. Methods / Functions

Both perform similar tasks.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a)
print(b)
```

# 1. UNION OPERATOR ( | )

Union combines all unique elements from both sets.

Duplicate values are automatically removed.

# RULES OF |

1. Used between two sets
2. Removes duplicate values
3. Returns new set

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a | b)
```

```python
x = {"python", "sql"}
y = {"sql", "excel"}

print(x | y)
```

# 2. INTERSECTION OPERATOR ( & )
Intersection returns only common elements present in both sets.

# RULES OF &

1. Returns common values only
2. Returns new set
3. Duplicate values ignored automatically

```python
a = {1, 2, 3, 4}
b = {3, 4, 5}

print(a & b)
```

```python
x = {"python", "sql"}
y = {"sql", "excel"}

print(x & y)
```

# 3. DIFFERENCE OPERATOR ( - )
Difference returns elements present in first set but not in second set.

# RULES OF -

1. Order matters
2. Returns unique values
3. Returns new set

```python
a = {1, 2, 3, 4}
b = {3, 4, 5}

print(a - b)
```

```python
print(b - a)
```

# 4. SYMMETRIC DIFFERENCE OPERATOR ( ^ )

Returns elements that are NOT common in both sets.

# RULES OF ^

1. Removes common values
2. Returns uncommon values
3. Returns new set

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a ^ b)
```

```python
x = {"python", "sql"}
y = {"sql", "excel"}

print(x ^ y)
```

# SET FUNCTIONS / METHODS

# 1. union()
union() combines all unique elements from sets.

Same work as | operator.

# RULES OF union()

1. Can take multiple arguments
2. Removes duplicate values
3. Returns new set
4. Original set does not change

```python
a = {1, 2}
b = {3, 4}

print(a.union(b))
```

```python
a = {1}
b = {2}
c = {3}

print(a.union(b, c))
```

# 2. difference()
difference() returns values present in first set but not in others.

# RULES OF difference()

1. Order matters
2. Can take multiple arguments
3. Returns new set
4. Original set does not change

```python
a = {1, 2, 3}
b = {2, 3}

print(a.difference(b))
```

```python
a = {1, 2, 3, 4}
b = {2}
c = {3}

print(a.difference(b, c))
```

# 3. difference_update()
difference_update() removes common elements directly from original set.

# RULES OF difference_update()

1. Modifies original set
2. Returns None
3. Can take multiple arguments

```python
a = {1, 2, 3}
b = {2}

a.difference_update(b)

print(a)
```

```python
a = {1, 2, 3, 4}
b = {2}
c = {3}

a.difference_update(b, c)

print(a)
```

# 4. intersection()
intersection() returns common elements from sets.
Same work as & operator.

# RULES OF intersection()

1. Can take multiple arguments
2. Returns common values only
3. Returns new set
4. Original set unchanged

```python
a = {1, 2, 3}
b = {2, 3, 4}

print(a.intersection(b))
```

```python
a = {1, 2, 3}
b = {2, 3}
c = {3}

print(a.intersection(b, c))
```

# 5. intersection_update()
intersection_update() keeps only common elements in original set.

# RULES OF intersection_update()

1. Modifies original set
2. Returns None
3. Can take multiple arguments

```python
a = {1, 2, 3}
b = {2, 3, 4}

a.intersection_update(b)

print(a)
```

```python
a = {1, 2, 3, 4}
b = {2, 3}
c = {3}

a.intersection_update(b, c)

print(a)
```

# 6. symmetric_difference()
Returns values that are not common in both sets.
Same work as ^ operator.

# RULES OF symmetric_difference()

1. Takes only ONE argument
2. Returns new set
3. Original set unchanged

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.symmetric_difference(b))
```

```python
x = {"python", "sql"}
y = {"sql", "excel"}

print(x.symmetric_difference(y))
```

# 7. symmetric_difference_update()
Updates original set with non-common values.

# RULES OF symmetric_difference_update()

1. Takes only ONE argument
2. Modifies original set
3. Returns None

```python
a = {1, 2, 3}
b = {3, 4, 5}

a.symmetric_difference_update(b)

print(a)
```

```python
x = {"python", "sql"}
y = {"sql", "excel"}

x.symmetric_difference_update(y)

print(x)
```

# 8. issubset()
Checks whether all elements of one set exist inside another set.

# RULES OF issubset()

1. Returns True or False
2. Takes only ONE argument
3. Smaller set checked inside bigger set

```python
a = {1, 2}
b = {1, 2, 3}

print(a.issubset(b))
```

```python
print(b.issubset(a))
```

# 9. issuperset()
Checks whether one set contains all elements of another set.

# RULES OF issuperset()

1. Returns True or False
2. Takes only ONE argument
3. Bigger set checked against smaller set

```python
a = {1, 2, 3}
b = {1, 2}

print(a.issuperset(b))
```

```python
print(b.issuperset(a))
```

# 10. isdisjoint()
Checks whether two sets have NO common elements.

# RULES OF isdisjoint()

1. Returns True or False
2. Takes only ONE argument
3. Returns False if even one common value exists

```python
a = {1, 2}
b = {3, 4}

print(a.isdisjoint(b))
```

```python
a = {1, 2}
b = {2, 3}

print(a.isdisjoint(b))
```
