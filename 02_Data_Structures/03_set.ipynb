# SET IN PYTHON

A set is a collection of unordered and unique elements in Python.

It is mainly used when we do not want duplicate values.

Sets are mutable, but they only store immutable elements inside them.

Important behavior:
- No duplicate values allowed
- Unordered (no fixed index)
- Fast for searching operations

Real-life use cases:
- removing duplicates from data
- membership testing
- mathematical set operations (union, intersection)

# RULES OF SET

1. Set is unordered (no index)
2. No duplicate values allowed. If given, it returns only one occurence of that value
3. Mutable (can be changed)
4. Elements must be immutable (no list inside set)
5. Written using {}
6. We can't use indexing and slicing

# IMPORTANT RULES / BEHAVIOR

1. {} empty braces is NOT a set → it is a dictionary
2. Empty set must be created using set()
3. Set = NOT index based structure
4. It uses hashing (internal system) So order is not guaranteed

Example:
{}       → dictionary
set()    → empty set

3. If all elements are same datatype, output may appear unordered but unique values only

```python
# Creating a set
s1 = {1, 2, 3, 4, 5}
print(s1)
```

```python
# duplicate values automatically removed
s2 = {1, 2, 2, 3, 3, 4}
print(s2)
```

```python
# empty set
s3 = set()
print(type(s3))
```

# SET METHODS

# add()

add() is used to insert a single element into a set.

# RULES OF ADD()

1. Adds only one element at a time
2. If element already exists, no change happens
3. Cannot add mutable types like list

```python
# Adding 4 in set
s = {1, 2, 3}

s.add(4)
print(s)
```

```python
# Duplicate values are not allowed
s.add(2) 
print(s)
```

# copy()

copy() creates a duplicate set with separate memory.

```python
# create a copy of set and add 10 in copied set and print both sets
s1 = {1, 2, 3}
s2 = s1.copy()

s2.add(10)

print(s1)
print(s2)
```

# clear()

clear() removes all elements from the set.

```python
# remove all values 
s = {1, 2, 3}

s.clear()
print(s)
```

# pop()

pop() removes a random element from the set because sets are unordered.

IMP 
---> This pop() is not like list pop()

It cannot removes a index based value like list pop()

```python
s = {10, 20, 30, 40}

s.pop()
print(s)
```

# discard()

discard() removes a specified element if it exists.

If element is not present, it does NOT give error.

```python
s = {1, 2, 3}

s.discard(2)
print(s)
```

```python
# 10 is not exist in set
s.discard(10)
print(s)
```

# remove()

remove() deletes a specified element from the set.

If element is not present, it gives error.

```python
s = {1, 2, 3}

s.remove(2)
print(s)
```

```python
# 10 is not exist in list
s.remove(10) 
```
