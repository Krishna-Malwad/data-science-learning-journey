-------
# range() in Python
-------

## What is range()?

range() is a built-in function used to generate a sequence of numbers.

It is mostly used with loops.

range() does NOT store all numbers directly.
It generates numbers when needed.

-----
 ## Syntax of range()

range(start, stop, step)

------
## Parameters

### start
Starting value of sequence

### stop
Ending value (not included)

### step
Increment / decrement value

----

## Important Rules of range()

1. stop value is NOT included
2. Default start value is 0
3. Default step value is 1
4. step cannot be 0
5. range() returns range object
6. Negative step is used for reverse counting
------

```python
# range(stop)

print(range(5))
```

-----
## Important Understanding

range(5) means:

start = 0  
stop = 5  
step = 1

---

```python
# converting range object into list

print(list(range(5)))
```

```python
# range(start, stop)

print(list(range(1, 6)))
```

```python
# range(start, stop, step)

print(list(range(1, 11, 2)))
```

```python
# reverse counting using negative step

print(list(range(10, 0, -1)))
```

-----
## Output Understanding

range(10, 0, -1)

start = 10  
stop = 0  
step = -1

Output:
10 → 9 → 8 → ... → 1

---

```python
# using range() with loop

for i in range(5):
    print(i)
```

-----
## Most Common Usage

range() is mostly used with:
- for loops
- indexing
- repetitions
- pattern problems
_____

## Printing range() Directly

range() does not directly show all values.

It returns a range object.

----

```python
# printing range object

a = range(10, 0, -1)

print(a)
```

-----
## Why list(), tuple(), or set() is Used

range() generates values lazily.

To see all generated values together,
we convert range object into:
- list
- tuple
- set
-----

```python
# converting range into list

print(list(range(5)))
```

```python
# converting range into tuple

print(tuple(range(5)))
```

```python
# converting range into set

print(set(range(5)))
```

-----
## Important Understanding

range() itself is NOT:
- list
- tuple
- set

It is a range object.

Values are generated only when needed.

-----
## Most Common Usage

### Use list(range())
When you want to see all values together.

### Use range() directly
Mostly inside loops.

----

## Accessing Values from range()

Values inside range() can be accessed using indexing.

-----

```python
# accessing values from range

r = range(5)

print(r[0])
print(r[3])
```

----
## Length of range()

len() returns total number of elements inside range.

-----

```python
# length of range

print(len(range(10)))
```

_______
## Membership Operator in range()

Membership operators are used to check whether a value exists in range or not.

_____

```python
# membership operator with range

print(5 in range(10))
print(20 in range(10))
```

-----
## Negative Ranges

range() can also generate negative numbers.

----

```python
# negative range

print(list(range(-5, 6)))
```

-----
## Reverse Counting using range()

Negative step is used for reverse counting.
    
----

```python
# reverse counting

print(list(range(10, 0, -1)))
```

```python
# reverse counting with custom step

print(list(range(20, 0, -2)))
```

----
## Empty Range (Very Important)

If direction and step do not match,
range() returns empty output.

-----

```python
# empty range example

print(list(range(1, 10, -1)))
```

---
## Why Empty Output?

range(1, 10, -1)

Python checks:

1 > 10

Condition becomes False immediately,
so no values are generated.

----

## How range() Works Internally

### Positive Step
Python checks:
current value < stop

### Negative Step
Python checks:
current value > stop

----

## Methods of range()

range object has only two methods:

1. count()
2. index()
-------

## count()

count() returns how many times a value exists in range.

----

```python
# count() method

r = range(1, 11)

print(r.count(5))
print(r.count(20))
```

------
## Important Note about count()

Since range contains unique values,
count() usually returns:

1 → if value exists  
0 → if value does not exist

-----

## index()

index() returns the position of a value in range.

----

```python
# index() method

r = range(10, 51, 10)

print(r.index(30))
```

---
## Important Rules of index()

1. Returns position/index
2. Value must exist
3. If value does not exist → ValueError
----


```python
# index() error example

r = range(1, 10)

print(r.index(20))
```

----
## Final Understanding

range() is mainly used for:
- loops
- repetitions
- indexing
- pattern problems

It generates values efficiently without storing all numbers in memory.

```python
# Create a range from 1 to 10 and print it as a list

# creating range from 1 to 10
a = range(1, 11)

# converting range into list
print(list(a))
```

```python
# Create a range from 5 to 20 with step 3 and print it as a list

# creating range with step 3
a = range(5, 21, 3)

# converting range into list
print(list(a))
```

```python
# Print numbers from 10 to 1 using range()

# creating reverse range
a = range(10, 0, -1)

# converting range into list
print(list(a))
```

```python
# Print all odd numbers from 1 to 15 using range()

# creating range for odd numbers
a = range(1, 16, 2)

# converting range into set
print(set(a))
```

```python
# Print all even numbers from 2 to 20 using range()

# creating range for even numbers
a = range(2, 21, 2)

# converting range into tuple
print(tuple(a))
```

```python
# Access the third element from range(1, 10)

# creating range
a = range(1, 10)

# printing third element
print(a[2])
```

```python
# Check whether 25 exists in range(1, 50, 5)

# creating range
a = range(1, 50, 5)

# checking membership
print(25 in a)
```

```python
# Find the length of range(1, 101)

# creating range
a = range(1, 101)

# finding length of range
print(len(a))
```

```python
# Count how many times 8 appears in range(1, 20)

# counting occurrence of 8
print(range(1, 20).count(8))
```

```python
# Find the index position of 40 in range(10, 51, 10)

# finding index position of 40
print(range(10, 51, 10).index(40))
```

```python
# Create a negative range from -10 to 0

# creating negative range
print(list(range(-10, 0)))
```

```python
# Create a reverse range from 50 to 10 with step -5

# creating reverse range
print(list(range(50, 10, -5)))
```

```python
# Create an empty range and print it as a list

# creating empty range
print(list(range(0)))
```

```python
# Print the last value of range(1, 20)

# printing last value from range
print(range(1, 20)[-1])
```

```python
# Print first five values generated by range(100, 200, 10)

# creating range
a = range(100, 200, 10)

# printing first five values
print(list(a)[:5])
```
