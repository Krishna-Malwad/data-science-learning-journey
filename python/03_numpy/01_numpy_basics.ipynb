# NumPy Basics

NumPy is one of the most important libraries in Python used for numerical computing and data science.

-----
## What is NumPy?

NumPy stands for **Numerical Python**.

It is a Python library used for working with:
- arrays
- matrices
- mathematical operations
- large datasets
----

## History of NumPy

NumPy was created in 2005 by Travis Oliphant.

It was built by combining:
- Numeric (older library)
- Numarray

NumPy became the standard library for numerical computing in Python.

---

## Why NumPy was created?

Before NumPy:
- Python lists were used for numbers
- They were slow for large data
- No mathematical operations directly

NumPy was created to solve these problems:
- Faster computation
- Less memory usage
- Easy mathematical operations

-----
## Why NumPy is better than Python lists?

Python list:
- slow for large data
- no built-in math operations

NumPy array:
- faster
- uses less memory
- supports vector operations
-----
## NumPy Array (Main Concept)

The most important feature of NumPy is the **array**.

NumPy array is similar to a list but:
- faster
- more powerful
- supports mathematical operations

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

----
## Types of NumPy Arrays

1D Array → single row  
2D Array → rows and columns (matrix)  
3D Array → collection of matrices

----
## Creating NumPy Arrays

We can create arrays using:
- np.array()
- np.arange()
- np.zeros()
- np.ones()
- random functions

---

## np.array()

np.array() is used to create a NumPy array from a Python list.

----

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

----
## np.arange()

np.arange() is used to create arrays with a range of values.

Syntax:
np.arange(start, stop, step)

---

```python
import numpy as np

print(np.arange(1, 10))
```

```python
print(np.arange(1, 10, 2))
```

----
## np.zeros()

np.zeros() creates an array filled with 0s.

---

```python
print(np.zeros(5))
```

-----
## np.ones()

np.ones() creates an array filled with 1s.

---

```python
print(np.ones(5))
```

----
## Random Functions in NumPy

NumPy provides random functions to generate random numbers.

### np.random.rand()

Generates random numbers between 0 and 1.

---

```python
print(np.random.rand(5))
```

----
### np.random.randint()

Generates random integers between a given range.

----

```python
print(np.random.randint(1, 10, 5))
```

----
## Random Functions in NumPy (Important Concept)

NumPy random functions work differently:

### np.random.rand()
- Generates random numbers between 0 and 1
- Output is float values
- You only control number of outputs

### np.random.randint()
- Generates random integers
- You can control:
  - start range
  - end range
  - number of values

### Key Difference:
- rand() → fixed range (0 to 1)
- randint() → custom range defined by user
----

## Indexing in NumPy Arrays

Indexing is used to access individual elements from a NumPy array.

Positive indexing starts from 0.

Negative indexing starts from -1.

-----

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

# accessing first element
print(arr[0])

# accessing third element
print(arr[2])

# accessing last element
print(arr[-1])
```

-----
## Slicing in NumPy Arrays

Slicing is used to access multiple elements from an array.

Syntax:
array[start : stop : step]

Important:
- stop index is excluded
----

```python
arr = np.array([10, 20, 30, 40, 50])

# slicing elements from index 1 to 3
print(arr[1:4])
```

```python
# elements from beginning to index 2
print(arr[:3])
```

```python
# elements from index 2 till end
print(arr[2:])
```

```python
# slicing with step
print(arr[::2])
```

----
## Important Difference

Indexing → returns single element

Slicing → returns multiple elements
