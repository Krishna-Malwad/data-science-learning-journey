# Dictionary in Python
----

Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as map or assosiative arrays

-----
syntax: 
dict = { 'name' : 'Krishna' , 'age' : 20 , 'gender' : 'male' }

----
# Characterstics:

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

----

# Rules of Dictionary
- Keys must be unique  
- Keys must be immutable  
- Values can be any type  
- Dictionary is ordered (Python 3.7+)

------

## Key and Value in Dictionary

A dictionary stores data in **key-value pairs**.

---

### Key:
- Acts as an identifier (like a label)
- Must be unique (no duplicate keys allowed)
- Must be immutable (string, int, tuple can be used as keys)
- Cannot be list, set, or dictionary

---

### Value:
- Data stored for a key
- Can be of any type (int, string, list, set, dict, etc.)
- Can be duplicated
- Mutable or immutable both are allowed

----

# # CREATE DICTIONARY

```python
# Creating a dictionary

student = {"name": "Krishna","age": 20,"course": "Data Science"}

print(student)
print(type(student))
```

# # ACCESSING VALUES

```python
# Accessing values using keys

student = {"name": "Krishna","age": 20}

print(student["name"])   # Accessing the value of "name" key
print(student["age"])    # Accessing the value of "age" key
```

```python
# If key does not exist → KeyError occurs

print(student)

print(student["Roll_No"])   # "Roll_No" key does not exist in student
```

----

ONE MORE METHOD TO ACCESS VALUES

-----

```python
# Accessing values using get() method

student = {"name": "Krishna", "age": 20}

print(student)              # prints full dictionary

print(student.get("name"))  # safely access value of key "name"

print(student.get("age"))   # safely access value of key "age"
```

----

# # ADDING KEY-VALUE PAIR

----

```python
# Adding key-value pair in dictionary

student = {"name": "Krishna","age": 20}

# adding new key-value pair
student["course"] = "Data Science"

print(student)
```

----

# # REMOVE KEY-VALUE PAIR

----

# 1. Using pop() method 
----
# About pop()

- Removes key-value pair using key
- Key must exist, otherwise KeyError occurs
- Returns removed value

-----

```python
# Remove a key-value pair using pop()

student = {"name": "Krishna", "age": 20,"course": "Data Science"}

student.pop("age")   # removes key "age" and its value

print(student)
```

-----
# 2. Using del keyword
----
# About del

- Deletes key-value pair directly
- No return value
- If key does not exist → KeyError
----

```python
# Remove a key-value pair using del

student = {"name": "Krishna", "age": 20, "course": "Data Science"}

del student["age"]   # deletes key "age"

print(student)
```

----
# # EDITING KEY–VALUE PAIR
----

# 1. Editing using direct key access
---
# About
- If key exists → value will be updated
- If key does not exist → new key-value pair will be added
----

```python
# Editing value of an existing key

student = {"name": "Krishna", "age": 20, "course": "Data Science"}

student["age"] = 21   # updating existing value

print(student)
```

----
# 2. Editing multiple values
----
About
- Dictionary does NOT allow duplicate keys
- So editing a key means overwriting its value

```python
# Updating multiple values

student = {"name": "Krishna", "age": 20, "course": "Data Science"}

# updating name value
student["name"] = "Kumar"

# updating course value
student["course"] = "AI/ML"

# printing final updated dictionary
print(student)
```

# # DICTIONARY METHODS
----


## 1. clear()

- The clear() method removes all key-value pairs from a dictionary.
- It makes the dictionary empty.

----

```python
# clear() 

student = {"name": "Krishna", "age": 20, "course": "Data Science"}

student.clear()

print(student)
```

---
## 2. copy()

- The copy() method creates a shallow copy of a dictionary.
- Changes in copied dictionary do NOT affect the original.
---

```python
# copy() 

student = {"name": "Krishna", "age": 20}

new_student = student.copy()  

print(new_student)
```

---
## 3. fromkeys()

- The fromkeys() method creates a new dictionary from given keys and assigns the same value to all keys.
---

```python
# fromkeys() 

keys = ["name", "age", "course"]

student = dict.fromkeys(keys, "unknown")

print(student)
```

---
## 4. get()

- The get() method returns the value of a key safely.
- If key is not found, it returns None instead of error.
---

```python
# get()

student = {"name": "Krishna", "age": 20}

print(student.get("name"))
print(student.get("marks"))
```

---
## 5. items()

- The items() method returns all key-value pairs as tuples.
----

```python
# items() 

student = {"name": "Krishna", "age": 20}

print(student.items())
```

----
## 6. keys()

- The keys() method returns all keys from the dictionary.
----

```python
# keys()

student = {"name": "Krishna", "age": 20}

print(student.keys())
```

----
## 7. pop()

- The pop() method removes a key-value pair using the key and returns its value.
----

```python
# pop() 

student = {"name": "Krishna", "age": 20}

student.pop("age")

print(student)
```

----
## 8. popitem()

- The popitem() method removes the last inserted key-value pair.
----

```python
# popitem()

student = {"name": "Krishna", "age": 20}

student.popitem()

print(student)
```

-----
## 9. setdefault()

- The setdefault() method returns value of a key if it exists.
- If not, it inserts the key with a default value.
-----

```python
# setdefault()

student = {"name": "Krishna", "age": 20}

student.setdefault("course", "Data Science")

print(student)
```

----
## 10. update()

- The update() method adds new key-value pairs or updates existing ones.
---

```python
# update()

student = {"name": "Krishna", "age": 20}

student.update({"course": "Data Science"})
student.update({"name":"Arjun"})

print(student)
```

----
## 11. values()

- The values() method returns all values from the dictionary.
---

```python
# values() method

student = {"name": "Krishna", "age": 20}

print(student.values())
```
