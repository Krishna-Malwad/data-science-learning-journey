# DATA TYPES IN PYTHON

```python
# Integer
a = 10
print(type(a))
```

```python
# Float
b = 10.5
print(type(b))
```

```python
# String
name = "Krushna"
print(type(name))
```

```python
# Boolean
is_student = True
print(type(is_student))
```

```python
# Complex
c = 2 + 3j
print(type(c))
```
