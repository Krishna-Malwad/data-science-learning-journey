# FUNCTIONS IN PYTHON

Functions are predefined or user-defined blocks of code that perform a specific task.

They are used to avoid repetition and make code reusable and organized.

Python provides built-in functions like print(), input(), id(), type().
We can also define our own functions using def keyword.

# RULES OF FUNCTIONS

1. Function must be called to execute
2. input() always returns string
3. type() returns data type of variable
4. id() returns memory address of object
5. print() displays output on screen

```python
name = input("Enter your name: ")
print(name)
```

INPUT() BYDEFAULT TAKES INPUT AS STRING 

```python
age = input("Enter age: ")
print(type(age))
```

```python
# Type casting with input

age = int(input("Enter age: "))
print(age)
print(type(age))
```

```python
a = 10
print(id(a))
```

# MODULES IN PYTHON

A module is a Python file that contains reusable code like functions, variables, or classes.

Modules help in organizing code into separate files instead of writing everything in one file.

Python has two types:
1. Built-in modules (math, random, keyword)
2. User-defined modules

# RULES OF MODULES

1. Must be imported using import keyword
2. Contains reusable code
3. Can be built-in or user-defined

# MATH MODULE

math module provides mathematical functions like square root, power, ceiling, floor, etc.

It is used for advanced mathematical calculations in Python.

```python
import math

print(math.sqrt(25))
print(math.pow(2, 3))
print(math.floor(4.7))
print(math.ceil(4.2))
```

# KEYWORD MODULE

keyword module is used to check Python reserved words.

Keywords are special words that cannot be used as variable names.

```python
import keyword

print(keyword.kwlist)
```

```python
import keyword

print(keyword.iskeyword("for"))
print(keyword.iskeyword("hello"))
```

# PACKAGES IN PYTHON

A package in Python is a way to organize multiple modules inside a folder.

Think of a real-world project:
- You cannot keep all code in one file
- So you split code into multiple modules (files)
- Then you group those modules inside a folder → this folder is called a PACKAGE

Example:
A calculator project can be split like:
- add.py (module)
- subtract.py (module)
- multiply.py (module)

All these modules are placed inside one folder called "calculator package"

So:

Function → inside Module → inside Package

Packages help in:
- large project organization
- clean structure
- reusability
- easy maintenance

# RULES OF PACKAGES

1. Package is a folder containing modules
2. Each module is a .py file
3. A package can contain multiple modules (fun, math, etc.)
4. Used in large projects for structure
5. Modules inside package are accessed using dot (.) notation

```python
# Example structure:

# calculator_package/
#     add.py
#     subtract.py
#     multiply.py
```

```python
# Take user input and show type and id
name = input("Enter your name: ")

print(name)
print(type(name))
print(id(name))
```

```python
# Take a number and print square root using math module
import math

num = int(input("Enter number: "))

print(math.sqrt(num))
```

```python
# Take two numbers and perform all basic math operations
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```
